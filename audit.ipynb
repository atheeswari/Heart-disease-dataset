```python
import pandas as pd


def check_chest_pain(df):
    # Filter for patients having chest pain (assuming column 'cp' > 0 indicates chest pain)
    has_cp = df["Chest_Pain"] > 0

    patients_with_cp = int(has_cp.sum())
    heart_disease_with_cp = int(((df["Heart_Risk"] == 1) & has_cp).sum())
    no_heart_disease_with_cp = int(((df["Heart_Risk"] == 0) & has_cp).sum())

    return {
        "patients_with_chest_pain": patients_with_cp,
        "heart_disease_with_chest_pain": heart_disease_with_cp,
        "no_heart_disease_with_chest_pain": no_heart_disease_with_cp,
    }


# Example function call:
result = check_chest_pain(df)
print(result)
```

```python
import pandas as pd

# 1. Define the missing helper function
def closed_rate_by(df, column_name, target_col="Heart_Disease"):
    """Calculates the mean target rate grouped by the given column."""
    return df.groupby(column_name)[target_col].mean()

# 2. Define the main calculation function
def calculate_ai_adoption_rates(df, column_name="AI_Adoption_Level"):
    order = ["None", "Exploratory", "Moderate", "Advanced", "AI-Native"]
    rates = closed_rate_by(df, column_name)
    
    result = {
        level: round(float(rates[level]), 2)
        for level in order
        if level in rates.index and pd.notna(rates[level])
    }
    
    if result:
        vals = list(result.values())
        result["spread_pp"] = round(max(vals) - min(vals), 2)
    else:
        result["spread_pp"] = 0.0
        
    return result

# 3. Sample DataFrame for testing
sample_data = {
    "Chest_Pain": ["None", "Exploratory", "Moderate", "Advanced", "AI-Native", "Moderate"],
    "Heart_Disease": [0, 1, 1, 0, 1, 0]
}
df = pd.DataFrame(sample_data)

# 4. Run function and print output
output = calculate_ai_adoption_rates(df, column_name="Chest_Pain")
print(output)
```

```python
import numpy as np
import pandas as pd


def compute_log_transformed_correlations(
    df: pd.DataFrame, pairs: list[tuple[str, str]], decimals: int = 3
) -> dict[str, float]:

    # Extract unique column names present across all input pairs
    unique_columns = {col for pair in pairs for col in pair}

    # Pre-calculate log1p arrays for all required columns
    logged_data = {
        col: np.log1p(df[col].to_numpy(dtype=float)) for col in unique_columns
    }

    # Calculate Pearson correlation coefficient for each pair
    correlations = {}
    for left, right in pairs:
        # Compute correlation matrix and extract off-diagonal value r_xy
        corr_matrix = np.corrcoef(logged_data[left], logged_data[right])
        r_val = float(corr_matrix[0, 1])

        correlations[f"{left}~{right}"] = round(r_val, decimals)

    return correlations


# --- Sample Data Setup & Execution ---
data = {
    "age": [25, 30, 35, 40, 45],
    "cholesterol": [180, 200, 220, 240, 260],
    "blood_pressure": [120, 125, 130, 135, 140],
}
df = pd.DataFrame(data)

pairs = [("age", "cholesterol"), ("age", "blood_pressure")]

result = compute_log_transformed_correlations(df, pairs)
print(result)
```

```python
import pandas as pd

# Load dataset
df = pd.read_csv("heart_disease_risk_dataset_earlymed.csv")

def analyze_dataset(df=None):
    data = df if df is not None else pd.read_csv("heart_disease_risk_dataset_earlymed.csv")

    diagnostics = {
        "shape": data.shape,
        "uniqueness": data.nunique().to_dict(),
        "missingness": data.isnull().sum().to_dict(),
        "cardinalities": {col: data[col].nunique() for col in data.columns},
        "risk_level": "High risk",
    }

    return diagnostics

result = analyze_dataset(df)
print(result)
```
